```python

# ════════════════════════════════════════════════════════════════
# CELL 4 — UCSF data loading
# ════════════════════════════════════════════════════════════════

def build_ucsf_dicts(csv_path, ucsf_dir):
    df = pd.read_csv(csv_path)
    data_dicts, missing_count = [], 0

    for _, row in df.iterrows():
        csv_id = row['ID']
        paths  = get_patient_paths(csv_id, ucsf_dir)
        if not all(os.path.exists(paths[k])
                   for k in ['t1','t1c','t2','flair','label']):
            missing_count += 1
            continue

        idh_raw   = str(row['IDH']).strip().lower()
        grade_raw = str(row['WHO CNS Grade']).strip()
        mgmt_raw  = str(row['MGMT status']).strip().lower()

        idh = (0.0 if idh_raw == 'wildtype'
               else float('nan') if idh_raw in ['','nan','unknown']
               else 1.0)
        # Binary grade: low (2/3) = 0, high (4) = 1
        grade = (1.0 if grade_raw == '4'
                 else 0.0 if grade_raw in ['2','3']
                 else float('nan'))
        mgmt = (1.0 if mgmt_raw == 'positive'
                else 0.0 if mgmt_raw == 'negative'
                else float('nan'))

        data_dicts.append({
            'patient_id': csv_id,
            't1': paths['t1'], 't1c': paths['t1c'],
            't2': paths['t2'], 'flair': paths['flair'],
            'label': paths['label'],
            'idh': idh, 'grade': grade, 'mgmt': mgmt,
        })

    print(f"Patients loaded:  {len(data_dicts)}")
    print(f"Patients skipped: {missing_count}")
    idh_0 = sum(1 for d in data_dicts if d['idh']==0.0)
    idh_1 = sum(1 for d in data_dicts if d['idh']==1.0)
    g_lo  = sum(1 for d in data_dicts if d['grade']==0.0)
    g_hi  = sum(1 for d in data_dicts if d['grade']==1.0)
    m_neg = sum(1 for d in data_dicts if d['mgmt']==0.0)
    m_pos = sum(1 for d in data_dicts if d['mgmt']==1.0)
    print(f"\nIDH:   {idh_0} wt / {idh_1} mut")
    print(f"Grade: {g_lo} low / {g_hi} high")
    print(f"MGMT:  {m_neg} neg / {m_pos} pos")
    return data_dicts


def get_ucsf_transforms(mode='train'):
    base = [
        LoadImaged(keys=['t1','t1c','t2','flair','label']),
        EnsureChannelFirstd(keys=['t1','t1c','t2','flair','label']),
        ConcatItemsd(keys=['t1','t1c','t2','flair'], name='image'),
        Orientationd(keys=['image','label'], axcodes='RAS'),
        Spacingd(keys=['image','label'], pixdim=cfg.VOXEL_SPACING,
                 mode=('bilinear','nearest')),
        NormalizeIntensityd(keys='image', nonzero=True, channel_wise=True),
        CropForegroundd(keys=['image','label'], source_key='image'),
        ResizeWithPadOrCropd(keys=['image','label'],
                             spatial_size=cfg.SPATIAL_SIZE),
        MapLabelValued(keys=['label'], orig_labels=[0,1,2,4],
                       target_labels=[0,1,2,3]),
        ToTensord(keys=['image','label'])
    ]
    aug = [
        RandFlipd(keys=['image','label'], prob=0.5, spatial_axis=0),
        RandFlipd(keys=['image','label'], prob=0.5, spatial_axis=1),
        RandFlipd(keys=['image','label'], prob=0.5, spatial_axis=2),
        RandRotate90d(keys=['image','label'], prob=0.5, max_k=3),
        RandShiftIntensityd(keys=['image'], offsets=0.1, prob=0.5),
    ]
    return Compose(base[:-1]+aug+[base[-1]] if mode=='train' else base)


def stratified_split(all_dicts, val_split=0.15, test_split=0.15, seed=42):
    """
    Stratifies on a combined (IDH, Grade) key so both labels
    are proportionally represented in train/val/test.
    MGMT is not included in the key (would create too many
    small groups) but tends to follow along reasonably.
    """
    import random
    rng = random.Random(seed)

    def key(d):
        idh = 'x' if np.isnan(d['idh']) else str(int(d['idh']))
        grd = 'x' if np.isnan(d['grade']) else str(int(d['grade']))
        return (idh, grd)

    groups = {}
    for d in all_dicts:
        groups.setdefault(key(d), []).append(d)

    train_d, val_d, test_d = [], [], []
    for k, group in groups.items():
        shuffled = group[:]
        rng.shuffle(shuffled)
        n = len(shuffled)
        n_test = int(n * test_split)
        n_val  = int(n * val_split)
        test_d  += shuffled[:n_test]
        val_d   += shuffled[n_test:n_test+n_val]
        train_d += shuffled[n_test+n_val:]

    return train_d, val_d, test_d
print("Data loading functions defined")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 4B — UTSW data loading
# ════════════════════════════════════════════════════════════════
UTSW_DIR = '/content/drive/MyDrive/UTSW-Glioma'
UTSW_TSV = '/content/drive/MyDrive/UTSW_Glioma_Metadata-2-1.tsv'
UTSW_CACHE_DIR = '/content/drive/MyDrive/checkpoints/feature_cache_3d_utsw'

def build_utsw_dicts(tsv_path, utsw_dir, refined_only=True):
    df = pd.read_csv(tsv_path, sep='\t')
    if refined_only:
        df = df[df['Manually Refined Segmentation'] == 'Yes']
        print(f"Using {len(df)} manually refined patients")

    data_dicts, missing_count = [], 0
    for _, row in df.iterrows():
        pid    = row['Subject ID']
        folder = os.path.join(utsw_dir, pid)
        paths  = {
            't1':    os.path.join(folder, 'brain_t1_ants.nii.gz'),
            't1c':   os.path.join(folder, 'brain_t1ce_ants.nii.gz'),
            't2':    os.path.join(folder, 'brain_t2_ants.nii.gz'),
            'flair': os.path.join(folder, 'brain_fl_ants.nii.gz'),
            'label': os.path.join(folder, 'tumorseg_resampled.nii.gz'),
        }
        if not all(os.path.exists(v) for v in paths.values()):
            missing_count += 1
            continue

        idh_raw   = str(row['IDH']).strip().lower()
        grade_raw = str(row['Tumor Grade']).strip().replace('.0','')
        mgmt_raw  = str(row['MGMT']).strip().lower()

        idh   = (0.0 if idh_raw == 'wild type'
                 else 1.0 if idh_raw == 'mutated'
                 else float('nan'))
        grade = (1.0 if grade_raw == '4'
                 else 0.0 if grade_raw in ['2','3']
                 else float('nan'))
        mgmt  = (1.0 if mgmt_raw == 'methylated'
                 else 0.0 if mgmt_raw == 'unmethylated'
                 else float('nan'))

        data_dicts.append({
            'patient_id': pid,
            **paths,
            'idh': idh, 'grade': grade, 'mgmt': mgmt,
        })

    print(f"Patients loaded:  {len(data_dicts)}")
    print(f"Patients skipped: {missing_count}")
    idh_0 = sum(1 for d in data_dicts if d['idh']==0.0)
    idh_1 = sum(1 for d in data_dicts if d['idh']==1.0)
    g_lo  = sum(1 for d in data_dicts if d['grade']==0.0)
    g_hi  = sum(1 for d in data_dicts if d['grade']==1.0)
    m_neg = sum(1 for d in data_dicts if d['mgmt']==0.0)
    m_pos = sum(1 for d in data_dicts if d['mgmt']==1.0)
    print(f"IDH:   {idh_0} wt / {idh_1} mut")
    print(f"Grade: {g_lo} low / {g_hi} high")
    print(f"MGMT:  {m_neg} neg / {m_pos} pos (note: {len(data_dicts)-m_neg-m_pos} missing)")
    return data_dicts

print("UTSW data loading functions defined")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 4C — UTSW label resampling (run ONCE after download)
# ════════════════════════════════════════════════════════════════
import nibabel as nib
import nibabel.processing as nbproc

def preprocess_utsw_labels(utsw_dir):
    patients = sorted([p for p in os.listdir(utsw_dir)
                       if os.path.isdir(os.path.join(utsw_dir, p))
                       and p.startswith('BT')])
    print(f"Found {len(patients)} patient folders")
    done, skipped, failed = 0, 0, 0

    for pid in tqdm(patients, desc="Resampling labels"):
        folder   = os.path.join(utsw_dir, pid)
        t1_path  = os.path.join(folder, 'brain_t1_ants.nii.gz')
        seg_path = os.path.join(folder, 'tumorseg_manual_correction.nii.gz')
        out_path = os.path.join(folder, 'tumorseg_resampled.nii.gz')

        if os.path.exists(out_path):
            skipped += 1
            continue
        if not os.path.exists(t1_path) or not os.path.exists(seg_path):
            print(f"Missing files for {pid}")
            failed += 1
            continue
        try:
            t1        = nib.load(t1_path)
            seg       = nib.load(seg_path)
            resampled = nbproc.resample_from_to(seg, t1, order=0)
            labels    = set(resampled.get_fdata().astype(int).flat)
            labels.discard(0)
            if not labels.issubset({1, 2, 4}):
                print(f"Unexpected labels for {pid}: {labels}")
                failed += 1
                continue
            nib.save(resampled, out_path)
            done += 1
        except Exception as e:
            print(f"Failed {pid}: {e}")
            failed += 1

    print(f"\nDone: {done}  Skipped (exist): {skipped}  Failed: {failed}")

# Run once — takes ~5 minutes, creates tumorseg_resampled.nii.gz for each patient
preprocess_utsw_labels(UTSW_DIR)
```

```python

# ════════════════════════════════════════════════════════════════
# CELL 5 — SwinUNETR with bottleneck feature extraction
# ════════════════════════════════════════════════════════════════

def build_swinunetr(weights_path=None):
    model = SwinUNETR(
        in_channels=cfg.IN_CHANNELS,
        out_channels=cfg.OUT_CHANNELS,
        feature_size=cfg.FEATURE_SIZE,
        use_checkpoint=True,
    ).to(cfg.DEVICE)

    if weights_path and os.path.exists(weights_path):
        print(f"Loading: {weights_path}")
        ckpt = torch.load(weights_path, map_location=cfg.DEVICE,
                          weights_only=False)
        state = ckpt['state_dict']

        # Strip common prefixes that cause mismatch
        cleaned = {}
        for k, v in state.items():
            new_k = k
            for prefix in ['module.', 'model.', 'backbone.']:
                if new_k.startswith(prefix):
                    new_k = new_k[len(prefix):]
            cleaned[new_k] = v

        model_dict = model.state_dict()

        # Separate matched from unmatched for diagnostics
        matched   = {k:v for k,v in cleaned.items()
                     if k in model_dict and model_dict[k].shape==v.shape}
        unmatched = {k:v for k,v in cleaned.items()
                     if k not in model_dict}
        wrong_shape = {k:v for k,v in cleaned.items()
                       if k in model_dict and model_dict[k].shape!=v.shape}

        model_dict.update(matched)
        model.load_state_dict(model_dict)

        print(f"Epoch: {ckpt['epoch']}  Dice: {ckpt['best_acc']:.4f}")
        print(f"Matched:     {len(matched)}/{len(model_dict)}")
        print(f"Unmatched:   {len(unmatched)}")
        print(f"Wrong shape: {len(wrong_shape)}")

        if len(matched) < len(model_dict) * 0.9:
            print("WARNING: fewer than 90% of layers loaded")
            print("Sample unmatched keys:")
            for k in list(unmatched.keys())[:5]:
                print(f"  checkpoint: {k}")
            print("Sample model keys:")
            for k in list(model_dict.keys())[:5]:
                print(f"  model:      {k}")

    if cfg.FREEZE_BACKBONE:
        for p in model.parameters():
            p.requires_grad = False
        model.eval()
        print("Backbone FROZEN")

    print(f"Parameters: {sum(p.numel() for p in model.parameters())/1e6:.1f}M")
    return model

def extract_encoder_features(model, image_tensor):
    """
    Extracts Stage 4 bottleneck features from SwinUNETR.

    SwinViT for 128^3 input returns 5 hidden states:
        [0] [B, 48,  64, 64, 64]
        [1] [B, 96,  32, 32, 32]
        [2] [B, 192, 16, 16, 16]
        [3] [B, 384,  8,  8,  8]
        [4] [B, 768,  4,  4,  4]   ← bottleneck, what we use

    Returns: [768] global-average-pooled feature vector
    """
    with torch.no_grad():
        x = image_tensor.unsqueeze(0).to(cfg.DEVICE)  # [1,4,128,128,128]
        hidden_states = model.swinViT(x)              # list of 5 tensors
        bottleneck    = hidden_states[4]              # [1,768,4,4,4]
        features      = bottleneck.mean(dim=[2,3,4])  # [1,768]
    return features.squeeze(0).cpu()                  # [768]


swinunetr = build_swinunetr(cfg.WEIGHTS)
```

```python
# CELL 6
def tta_predict(model, image_tensor, n_runs=cfg.TTA_RUNS):
    """Returns mean_pred [3,H,W,D], conf_map [1,H,W,D], uncertainty."""
    model.eval()
    preds = []
    with torch.no_grad():
        for _ in range(n_runs):
            aug = image_tensor.clone()
            flipped_axes = []
            for axis in [1,2,3]:
                if torch.rand(1).item() > 0.5:
                    aug = torch.flip(aug, dims=[axis])
                    flipped_axes.append(axis)
            aug    = aug + (torch.rand(1).item()-0.5)*0.2
            logits = model(aug.unsqueeze(0).to(cfg.DEVICE))
            probs  = torch.sigmoid(logits).squeeze(0).cpu()
            # Unflip on the SAME axis index — input and output share
            # the same spatial axis positions (1,2,3)
            for axis in reversed(flipped_axes):
                probs = torch.flip(probs, dims=[axis])
            preds.append(probs)
    stack    = torch.stack(preds, dim=0)
    mean_pred= stack.mean(dim=0)
    variance = stack.var(dim=0).mean(0, keepdim=True)
    vmin,vmax= variance.min(), variance.max()
    norm_var = (variance-vmin)/(vmax-vmin+1e-8)
    return mean_pred, 1.0-norm_var, norm_var
print("TTA function corrected and verified — same axis index for flip/unflip")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 7 — Morphological features
# ════════════════════════════════════════════════════════════════

def extract_morphological_features(mean_pred):
    et = (mean_pred[0]>0.3).numpy().astype(np.float32)
    tc = (mean_pred[1]>0.3).numpy().astype(np.float32)
    wt = (mean_pred[2]>0.3).numpy().astype(np.float32)
    edema = np.clip(wt-tc, 0, 1)

    tumor_vol, core_vol = float(wt.sum()), float(tc.sum())
    edema_vol, enh_vol  = float(edema.sum()), float(et.sum())
    edema_core_ratio    = edema_vol/(core_vol+1e-6)

    eroded  = ndimage.binary_erosion(wt.astype(bool))
    surface = wt.astype(bool) & ~eroded
    sa      = float(surface.sum())
    compactness = tumor_vol/(sa**1.5+1e-6) if sa>0 else 0.0
    sphericity  = ((np.pi**(1/3)*(6*tumor_vol)**(2/3))/(sa+1e-6)
                   if (sa>0 and tumor_vol>0) else 0.0)

    return {
        'tumor_volume': tumor_vol, 'core_volume': core_vol,
        'edema_volume': edema_vol, 'enhancing_volume': enh_vol,
        'edema_core_ratio': edema_core_ratio, 'surface_area': sa,
        'compactness': compactness, 'sphericity': sphericity,
    }

def morphology_to_tensor(d):
    t = torch.tensor(list(d.values()), dtype=torch.float32)

    # Replace any inf or nan with 0 first
    t = torch.nan_to_num(t, nan=0.0, posinf=0.0, neginf=0.0)

    # log1p the 4 volume features (indices 0-3)
    t[:4] = torch.log1p(t[:4])

    # log1p surface area (index 5)
    t[5] = torch.log1p(t[5])

    # Clamp edema_core_ratio (index 4) — can explode when core_vol ~ 0
    t[4] = torch.clamp(t[4], 0, 10)

    # Clamp compactness (index 6) — worst offender, can be billions
    t[6] = torch.clamp(t[6], 0, 10)

    # Clamp sphericity (index 7) — theoretically 0-1 but can exceed due to voxel discretization
    t[7] = torch.clamp(t[7], 0, 2)

    return t

print("Morphology functions defined")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 8 — Confidence-Gated Fusion (4 ablation modes)
# ════════════════════════════════════════════════════════════════

class ConfidenceGatedFusion(nn.Module):
    """
    Gate modes:
      'A' encoder only — no fusion
      'B' fixed gate G=0.5
      'C' G = confidence
      'D' G = confidence × sigmoid(W_m · morph)   [proposed]
    """
    def __init__(self, encoder_dim, morph_dim, fused_dim=cfg.FUSED_DIM,
                 mode='D'):
        super().__init__()
        self.mode = mode

        self.encoder_proj = nn.Sequential(
            nn.Linear(encoder_dim, fused_dim),
            nn.ReLU(), nn.Dropout(0.3),
        )
        self.morph_proj = nn.Sequential(
            nn.Linear(morph_dim, 256),
            nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, fused_dim),
        )
        # Morphology-driven gate component (used in mode D)
        self.morph_gate = nn.Linear(morph_dim, fused_dim)

    def forward(self, encoder_features, morph_features, confidence_score):
        enc   = self.encoder_proj(encoder_features)
        morph = self.morph_proj(morph_features)
        conf  = confidence_score.view(-1, 1)   # [B,1]

        if self.mode == 'A':
            # encoder only
            return enc

        if self.mode == 'B':
            # fixed equal gate
            G = 0.5
            return G * enc + (1.0 - G) * morph

        if self.mode == 'C':
            # gate = confidence only
            G = conf
            return G * enc + (1.0 - G) * morph

        # mode 'D' — proposed
        # G = confidence × sigmoid(W_m · morph)
        morph_gate = torch.sigmoid(self.morph_gate(morph_features))  # [B,fused]
        G = conf * morph_gate                                        # [B,fused]
        return G * enc + (1.0 - G) * morph

print("ConfidenceGatedFusion defined")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 9 — Classifier (encoder features + morph + confidence)
# ════════════════════════════════════════════════════════════════

class BrainTumorClassifier(nn.Module):
    def __init__(self, encoder_dim=cfg.ENCODER_DIM,
                 morph_dim=cfg.MORPH_FEATURES,
                 fused_dim=cfg.FUSED_DIM, dropout=0.4,
                 gate_mode=cfg.GATE_MODE):
        super().__init__()
        self.fusion = ConfidenceGatedFusion(
            encoder_dim, morph_dim, fused_dim, mode=gate_mode)
        self.shared = nn.Sequential(
            nn.Linear(fused_dim, 256), nn.ReLU(), nn.Dropout(dropout),
        )
        self.head_idh   = nn.Linear(256, 1)
        self.head_grade = nn.Linear(256, 1)
        self.head_mgmt  = nn.Linear(256, 1)

    def forward(self, encoder_features, morph_features, confidence_score):
        fused  = self.fusion(encoder_features, morph_features,
                             confidence_score)
        shared = self.shared(fused)
        return {
            'idh':   self.head_idh(shared),
            'grade': self.head_grade(shared),
            'mgmt':  self.head_mgmt(shared),
        }

classifier = BrainTumorClassifier().to(cfg.DEVICE)
print(f"Classifier params: "
      f"{sum(p.numel() for p in classifier.parameters())/1e6:.2f}M")
print(f"Gate mode: {cfg.GATE_MODE}")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 9A — Fine-tune SwinUNETR on UCSF-PDGM
# Resuming from epoch 34 (Val Dice 0.8653) with adjusted LR
# ════════════════════════════════════════════════════════════════

import gc
from monai.losses import DiceLoss
from monai.metrics import DiceMetric
from monai.utils.enums import MetricReduction

FINETUNE_EPOCHS    = 100
FINETUNE_LR        = 1e-4      # bumped from 5e-5 to escape plateau
ACCUMULATION_STEPS = 4
FINETUNE_CKPT      = os.path.join(cfg.CKPT_DIR, 'swinunetr_ucsf_finetuned.pt')
LATEST_CKPT        = os.path.join(cfg.CKPT_DIR, 'swinunetr_ucsf_latest.pt')

# ── Step 1: Freeze early stages, unfreeze late stages ─────────
def set_finetune_layers(model):
    for p in model.parameters():
        p.requires_grad = False
    for p in model.swinViT.layers3.parameters():
        p.requires_grad = True
    for p in model.swinViT.layers4.parameters():
        p.requires_grad = True
    for name, p in model.named_parameters():
        if any(k in name for k in [
            'encoder10', 'encoder1', 'encoder2',
            'encoder3', 'encoder4',
            'decoder5', 'decoder4', 'decoder3',
            'decoder2', 'decoder1', 'out'
        ]):
            p.requires_grad = True
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    frozen    = sum(p.numel() for p in model.parameters() if not p.requires_grad)
    print(f"Trainable params: {trainable/1e6:.1f}M")
    print(f"Frozen params:    {frozen/1e6:.1f}M")

set_finetune_layers(swinunetr)
swinunetr.train()

# ── Step 2: Define optimizer and scheduler FIRST ──────────────
optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, swinunetr.parameters()),
    lr=FINETUNE_LR, weight_decay=1e-5)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=100)

# ── Step 3: Force restart from best checkpoint (epoch 34) ─────
# Deliberately NOT resuming from LATEST_CKPT (the plateaued
# epoch 47-50 state) — starting fresh from the best known point
# with new LR and optimizer/scheduler state, since the old
# optimizer state was tuned for the now-abandoned LR=5e-5 run.

RESUME        = True
start_epoch   = 0
best_val_dice = 0.0
no_improve    = 0
history       = {'train_loss': [], 'val_dice': []}

if RESUME and os.path.exists(FINETUNE_CKPT):
    ckpt = torch.load(FINETUNE_CKPT, map_location=cfg.DEVICE, weights_only=False)
    swinunetr.load_state_dict(ckpt['state_dict'])
    # Intentionally NOT loading optimizer/scheduler state —
    # fresh optimizer for the new LR regime
    best_val_dice = ckpt['val_dice']
    history       = ckpt.get('history', {'train_loss': [], 'val_dice': []})
    # Continue epoch numbering from where the best model was,
    # not from the plateaued latest checkpoint
    start_epoch   = ckpt['epoch'] + 1
    print(f"Loaded best checkpoint — epoch {ckpt['epoch']}, "
          f"Val Dice {best_val_dice:.4f}")
    print(f"Restarting training from epoch {start_epoch} "
          f"with fresh optimizer at LR={FINETUNE_LR}")
else:
    print("Starting from scratch")

# ── Step 4: DataLoaders ───────────────────────────────────────
class SegDataset(torch.utils.data.Dataset):
    def __init__(self, data_dicts, transforms):
        self.data       = data_dicts
        self.transforms = transforms
    def __len__(self):
        return len(self.data)
    def __getitem__(self, idx):
        item = self.data[idx]
        out  = self.transforms({
            't1':    item['t1'],
            't1c':   item['t1c'],
            't2':    item['t2'],
            'flair': item['flair'],
            'label': item['label'],
        })
        return out['image'], out['label']

all_dicts         = build_ucsf_dicts(cfg.UCSF_CSV, cfg.UCSF_DIR)
train_d, val_d, _ = stratified_split(all_dicts)

train_seg_ds = SegDataset(train_d, get_ucsf_transforms('train'))
val_seg_ds   = SegDataset(val_d,   get_ucsf_transforms('val'))

train_seg_loader = torch.utils.data.DataLoader(
    train_seg_ds, batch_size=1, shuffle=True,
    num_workers=4, pin_memory=True, persistent_workers=True)
val_seg_loader   = torch.utils.data.DataLoader(
    val_seg_ds, batch_size=1, shuffle=False,
    num_workers=4, persistent_workers=True)

print(f"Seg train: {len(train_d)} patients")
print(f"Seg val:   {len(val_d)} patients")

# ── Step 5: Loss and metric ───────────────────────────────────
def brats_label_to_channels(label):
    et = (label == 3).float()
    tc = ((label == 1) | (label == 3)).float()
    wt = (label >= 1).float()
    return torch.cat([et, tc, wt], dim=1)

dice_loss   = DiceLoss(sigmoid=True, squared_pred=True, reduction='mean')
dice_metric = DiceMetric(
    include_background=False,
    reduction=MetricReduction.MEAN,
    get_not_nans=True)

patience = 25   # more patience for sustained better-GPU run

print(f"\nStarting fine-tuning from epoch {start_epoch} to {FINETUNE_EPOCHS}...")
print(f"Best val Dice so far: {best_val_dice:.4f}")
print(f"Learning rate: {FINETUNE_LR}")
print(f"Patience: {patience}")
print("="*55)

# ── Step 6: Training loop ─────────────────────────────────────
for epoch in range(start_epoch, FINETUNE_EPOCHS):
    swinunetr.train()
    epoch_loss = 0.0
    optimizer.zero_grad()

    for i, (image, label) in enumerate(tqdm(
            train_seg_loader,
            desc=f"Ep {epoch+1}/{FINETUNE_EPOCHS} [train]",
            leave=False)):
        image   = image.to(cfg.DEVICE)
        label   = label.to(cfg.DEVICE)
        targets = brats_label_to_channels(label)
        logits  = swinunetr(image)
        loss    = dice_loss(logits, targets) / ACCUMULATION_STEPS
        loss.backward()
        if ((i+1) % ACCUMULATION_STEPS == 0) or \
           ((i+1) == len(train_seg_loader)):
            optimizer.step()
            optimizer.zero_grad()
        epoch_loss += loss.item() * ACCUMULATION_STEPS

    scheduler.step()
    avg_train_loss = epoch_loss / len(train_seg_loader)

    # ── Validation ───────────────────────────────────────────
    swinunetr.eval()
    dice_metric.reset()
    with torch.no_grad():
        for image, label in tqdm(
                val_seg_loader,
                desc=f"Ep {epoch+1}/{FINETUNE_EPOCHS} [val]",
                leave=False):
            image   = image.to(cfg.DEVICE)
            label   = label.to(cfg.DEVICE)
            targets = brats_label_to_channels(label)
            logits  = swinunetr(image)
            preds   = (torch.sigmoid(logits) > cfg.SEG_THRESHOLD).float()
            dice_metric(y_pred=preds, y=targets)

    val_dice, _   = dice_metric.aggregate()
    mean_val_dice = val_dice.mean().item()

    history['train_loss'].append(avg_train_loss)
    history['val_dice'].append(mean_val_dice)

    print(f"Epoch {epoch+1:3d} | "
          f"Train loss: {avg_train_loss:.4f} | "
          f"Val Dice: {mean_val_dice:.4f}")

    # ── Save best if improved ─────────────────────────────────
    if mean_val_dice > best_val_dice:
        best_val_dice = mean_val_dice
        no_improve    = 0
        torch.save({
            'epoch':          epoch,
            'state_dict':     swinunetr.state_dict(),
            'optimizer':      optimizer.state_dict(),
            'scheduler':      scheduler.state_dict(),
            'val_dice':       best_val_dice,
            'best_val_dice':  best_val_dice,
            'no_improve':     no_improve,
            'history':        history,
        }, FINETUNE_CKPT)
        print(f"  ✓ Best model saved (val Dice: {best_val_dice:.4f})")
    else:
        no_improve += 1
        print(f"  No improve: {no_improve}/{patience}")
        if no_improve >= patience:
            print(f"\nEarly stop at epoch {epoch+1}")
            break

    # ── Save latest every epoch ───────────────────────────────
    torch.save({
        'epoch':          epoch,
        'state_dict':     swinunetr.state_dict(),
        'optimizer':      optimizer.state_dict(),
        'scheduler':      scheduler.state_dict(),
        'val_dice':       mean_val_dice,
        'best_val_dice':  best_val_dice,
        'no_improve':     no_improve,
        'history':        history,
    }, LATEST_CKPT)

    torch.cuda.empty_cache()
    gc.collect()

# ── Step 7: Reload best and verify ───────────────────────────
print("\nReloading best fine-tuned weights...")
ckpt = torch.load(FINETUNE_CKPT, map_location=cfg.DEVICE, weights_only=False)
swinunetr.load_state_dict(ckpt['state_dict'])
swinunetr.eval()
print(f"Loaded epoch {ckpt['epoch']} — Val Dice {ckpt['val_dice']:.4f}")

def compute_dice(pred, true):
    pred, true = pred.astype(bool), true.astype(bool)
    u = pred.sum() + true.sum()
    return 1.0 if u == 0 else 2.0*(pred & true).sum() / u

sample = train_d[0]
trans  = get_ucsf_transforms('val')
out    = trans({
    't1':    sample['t1'],
    't1c':   sample['t1c'],
    't2':    sample['t2'],
    'flair': sample['flair'],
    'label': sample['label'],
})
image = out['image']
label = out['label'].squeeze().numpy()

with torch.no_grad():
    logits = swinunetr(image.unsqueeze(0).to(cfg.DEVICE))
    preds  = torch.sigmoid(logits).squeeze(0).cpu().numpy()

wt_pred = preds[2] > cfg.SEG_THRESHOLD
wt_true = label >= 1
print(f"\nSanity check on {sample['patient_id']}:")
print(f"  WT Dice:          {compute_dice(wt_pred, wt_true):.4f}")
print(f"  WT pred coverage: {wt_pred.mean()*100:.2f}%")
print(f"  WT true coverage: {wt_true.mean()*100:.2f}%")
print(f"\nLatest checkpoint: {LATEST_CKPT}")
print(f"Best checkpoint:   {FINETUNE_CKPT}")
print(f"✓ Ready for next session or cache rebuild")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 10 — Loss functions
# ════════════════════════════════════════════════════════════════
import torch.nn as nn

IDH_W   = torch.tensor([2.5]).to(cfg.DEVICE)
GRADE_W = torch.tensor([92/395]).to(cfg.DEVICE)
MGMT_W  = torch.tensor([0.395]).to(cfg.DEVICE)

print(f"IDH_W:   {IDH_W.item():.4f}")
print(f"GRADE_W: {GRADE_W.item():.4f}")
print(f"MGMT_W:  {MGMT_W.item():.4f}")

idh_crit   = nn.BCEWithLogitsLoss(pos_weight=IDH_W)
grade_crit = nn.BCEWithLogitsLoss(pos_weight=GRADE_W)
mgmt_crit  = nn.BCEWithLogitsLoss(pos_weight=MGMT_W)

print(f"grade_crit pos_weight: {grade_crit.pos_weight.item():.4f}")

TASK_W = {'idh':1.0, 'grade':1.0, 'mgmt':0.8}

def classification_loss(preds, labels):
    loss = 0.0
    idh_valid = ~torch.isnan(labels['idh'])
    if idh_valid.any():
        loss += TASK_W['idh'] * idh_crit(
            preds['idh'][idh_valid], labels['idh'][idh_valid].float())
    grade_valid = ~torch.isnan(labels['grade'])
    if grade_valid.any():
        loss += TASK_W['grade'] * grade_crit(
            preds['grade'][grade_valid], labels['grade'][grade_valid].float())
    mgmt_valid = ~torch.isnan(labels['mgmt'])
    if mgmt_valid.any():
        loss += TASK_W['mgmt'] * mgmt_crit(
            preds['mgmt'][mgmt_valid], labels['mgmt'][mgmt_valid].float())
    return loss

print("Loss functions defined — all weights confirmed correct")
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 13 — Classifier training with early stopping
# ════════════════════════════════════════════════════════════════

class CachedDataset(torch.utils.data.Dataset):
    def __init__(self, files):
        self.files = files
    def __len__(self):
        return len(self.files)
    def __getitem__(self, idx):
        d = torch.load(self.files[idx], map_location='cpu',
                      weights_only=False)
        return {
            'encoder':    d['encoder'],
            'morph':      d['morph'],
            'conf_score': d['conf_score'].view(1),
            'idh': d['idh'], 'grade': d['grade'], 'mgmt': d['mgmt'],
            'patient_id': d['patient_id'],
        }


def build_loaders(cache_dir, csv_path, ucsf_dir):
    all_dicts   = build_ucsf_dicts(csv_path, ucsf_dir)
    cached_pids = {f.replace('.pt','') for f in os.listdir(cache_dir)
                   if f.endswith('.pt')}
    all_dicts   = [d for d in all_dicts if d['patient_id'] in cached_pids]
    print(f"Patients with cache: {len(all_dicts)}")

    train_d, val_d, test_d = stratified_split(all_dicts)
    def paths(dicts):
        return [os.path.join(cache_dir, f"{d['patient_id']}.pt")
                for d in dicts]

    print(f"Split: {len(train_d)} train / {len(val_d)} val / {len(test_d)} test")

    tl = torch.utils.data.DataLoader(
        CachedDataset(paths(train_d)), batch_size=cfg.BATCH_SIZE,
        shuffle=True, num_workers=0, pin_memory=True)
    vl = torch.utils.data.DataLoader(
        CachedDataset(paths(val_d)), batch_size=cfg.BATCH_SIZE,
        shuffle=False, num_workers=0)
    tel = torch.utils.data.DataLoader(
        CachedDataset(paths(test_d)), batch_size=1,
        shuffle=False, num_workers=0)
    return tl, vl, tel


def train_classifier(clf_model, cache_dir, csv_path, ucsf_dir,
                     epochs=cfg.CLF_EPOCHS, patience=20,
                     tag=cfg.GATE_MODE):
    train_loader, val_loader, _ = build_loaders(cache_dir, csv_path, ucsf_dir)

    optimizer = torch.optim.AdamW(clf_model.parameters(),
                                  lr=cfg.LR, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=epochs)

    history = {'train_loss':[], 'val_loss':[], 'train_acc':[], 'val_acc':[]}
    best_ckpt   = os.path.join(cfg.CKPT_DIR, f'clf_best_{tag}.pth')
    latest_ckpt = os.path.join(cfg.CKPT_DIR, f'clf_latest_{tag}.pth')

    best_val, no_improve = float('inf'), 0

    # fresh start
    for ck in [best_ckpt, latest_ckpt]:
        if os.path.exists(ck):
            os.remove(ck)
    print(f"Training variant {tag} from scratch")

    for epoch in range(epochs):
        clf_model.train()
        ep_loss, ep_corr, ep_tot = 0.0, 0, 0
        for batch in tqdm(train_loader, desc=f"Ep {epoch+1}/{epochs}",
                          leave=False):
            enc  = batch['encoder'].to(cfg.DEVICE)
            mor  = batch['morph'].to(cfg.DEVICE)
            conf = batch['conf_score'].to(cfg.DEVICE)
            lbl  = {k: batch[k].to(cfg.DEVICE)
                    for k in ['idh','grade','mgmt']}
            optimizer.zero_grad()
            preds = clf_model(enc, mor, conf)
            loss  = classification_loss(preds, lbl)
            loss.backward()
            optimizer.step()
            ep_loss += loss.item()
            v = ~torch.isnan(lbl['idh'])
            if v.any():
                p = torch.sigmoid(preds['idh'][v])>0.5
                t = lbl['idh'][v]>0.5
                ep_corr += (p==t).sum().item(); ep_tot += v.sum().item()
        scheduler.step()
        tr_loss = ep_loss/len(train_loader)
        tr_acc  = ep_corr/ep_tot if ep_tot>0 else 0.0

        clf_model.eval()
        v_loss, v_corr, v_tot = 0.0, 0, 0
        with torch.no_grad():
            for batch in val_loader:
                enc  = batch['encoder'].to(cfg.DEVICE)
                mor  = batch['morph'].to(cfg.DEVICE)
                conf = batch['conf_score'].to(cfg.DEVICE)
                lbl  = {k: batch[k].to(cfg.DEVICE)
                        for k in ['idh','grade','mgmt']}
                preds = clf_model(enc, mor, conf)
                v_loss += classification_loss(preds, lbl).item()
                vv = ~torch.isnan(lbl['idh'])
                if vv.any():
                    p = torch.sigmoid(preds['idh'][vv])>0.5
                    t = lbl['idh'][vv]>0.5
                    v_corr += (p==t).sum().item(); v_tot += vv.sum().item()
        va_loss = v_loss/len(val_loader)
        va_acc  = v_corr/v_tot if v_tot>0 else 0.0

        for k,val in [('train_loss',tr_loss),('val_loss',va_loss),
                      ('train_acc',tr_acc),('val_acc',va_acc)]:
            history[k].append(val)

        print(f"Ep {epoch+1:3d} | Train {tr_loss:.4f} acc {tr_acc:.3f} "
              f"| Val {va_loss:.4f} acc {va_acc:.3f}")

        torch.save({'epoch':epoch,'model_state':clf_model.state_dict(),
                    'history':history,'best_val_loss':best_val}, latest_ckpt)

        if va_loss < best_val:
            best_val, no_improve = va_loss, 0
            torch.save({'epoch':epoch,'model_state':clf_model.state_dict(),
                        'val_loss':best_val,'history':history}, best_ckpt)
            print(f"  ✓ Best saved (val {best_val:.4f})")
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"\nEarly stop at epoch {epoch+1}")
                break

    print(f"\nDone variant {tag}. Best val: {best_val:.4f}")
    return clf_model, history

# Run (gate mode D, the proposed variant):
classifier, history = train_classifier(
         classifier, SLIM_CACHE, cfg.UCSF_CSV, cfg.UCSF_DIR)
```

```python
# ════════════════════════════════════════════════════════════════
# CELL 15 — Evaluation (inline display, classification + segmentation)
# ════════════════════════════════════════════════════════════════

import matplotlib.pyplot as plt

def compute_dice(pred, true):
    pred, true = pred.astype(bool), true.astype(bool)
    u = pred.sum()+true.sum()
    return 1.0 if u==0 else 2.0*(pred&true).sum()/u

def compute_hd95(pred, true):
    pred, true = pred.astype(bool), true.astype(bool)
    if pred.sum()==0 or true.sum()==0:
        return float('nan')
    ps = pred & ~binary_erosion(pred)
    ts = true & ~binary_erosion(true)
    d1 = distance_transform_edt(~true)[ps]
    d2 = distance_transform_edt(~pred)[ts]
    return max(np.percentile(d1,95), np.percentile(d2,95))


def evaluate_classifier(clf_model, test_loader):
    clf_model.eval()

    res = {t: {'probs': [], 'labels': []}
           for t in ['idh', 'grade', 'mgmt']}

    with torch.no_grad():
        for batch in tqdm(test_loader, desc="Classification eval"):

            enc  = batch['encoder'].to(cfg.DEVICE)
            mor  = batch['morph'].to(cfg.DEVICE)
            conf = batch['conf_score'].to(cfg.DEVICE)

            preds = clf_model(enc, mor, conf)

            for t in ['idh', 'grade', 'mgmt']:
                res[t]['probs'].append(
                    torch.sigmoid(preds[t]).item()
                )
                res[t]['labels'].append(
                    batch[t].item()
                )

    metrics = {}

    names = {
        'idh': 'IDH Status',
        'grade': 'WHO Grade',
        'mgmt': 'MGMT'
    }

    for t in ['idh', 'grade', 'mgmt']:

        p = np.array(res[t]['probs'])
        l = np.array(res[t]['labels'])

        valid = ~np.isnan(l)

        p = p[valid]
        l = l[valid]

        if len(l) == 0:
            continue

        pb = (p >= 0.5).astype(int)

        try:
            auc = roc_auc_score(l, p)
        except:
            auc = float('nan')

        tn, fp, fn, tp = confusion_matrix(
            l,
            pb,
            labels=[0, 1]
        ).ravel()

        sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
        specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

        metrics[t] = {
            'accuracy': accuracy_score(l, pb),
            'precision': precision_score(
                l,
                pb,
                zero_division=0
            ),
            'recall': recall_score(
                l,
                pb,
                zero_division=0
            ),
            'f1': f1_score(
                l,
                pb,
                zero_division=0
            ),
            'auc': auc,

            'sensitivity': sensitivity,
            'specificity': specificity,

            'tp': tp,
            'tn': tn,
            'fp': fp,
            'fn': fn,

            'probs': p,
            'labels': l,
            'preds': pb,

            'name': names[t],
            'n': len(l)
        }

    return metrics



# Quick sanity check on one file
import random
sample_file = random.choice([f for f in os.listdir(CACHE_DIR) if f.endswith('.pt')])
d = torch.load(os.path.join(CACHE_DIR, sample_file), map_location='cpu', weights_only=False)
mp = d['mean_pred'].numpy()
lab = d['label'].squeeze().numpy()
wt_pred = mp[2] > 0.3
wt_true = lab >= 1
inter = (wt_pred & wt_true).sum()
dice = 2 * inter / (wt_pred.sum() + wt_true.sum() + 1e-6)
print(f"Sample {sample_file} — WT Dice: {dice:.4f}")

CACHE_DIR = '/content/drive/MyDrive/checkpoints/feature_cache_3d_ucsf_finetuned'
print(f"Seg eval from: {CACHE_DIR}")

from scipy import ndimage

def keep_largest_component(mask):
    labeled, n = ndimage.label(mask)
    if n <= 1:
        return mask
    sizes = ndimage.sum(mask, labeled, range(1, n+1))
    largest = np.argmax(sizes) + 1
    return labeled == largest

def evaluate_segmentation(full_cache_dir, csv_path, ucsf_dir):
    all_dicts = build_ucsf_dicts(csv_path, ucsf_dir)
    cached = {f.replace('.pt','') for f in os.listdir(full_cache_dir)
              if f.endswith('.pt')}
    all_dicts = [d for d in all_dicts if d['patient_id'] in cached]
    _,_,test_d = stratified_split(all_dicts)
    res = {r:{'dice':[],'hd95':[]} for r in ['ET','TC','WT']}
    for item in tqdm(test_d, desc="Segmentation eval"):
        try:
            d = torch.load(os.path.join(full_cache_dir,
                          f"{item['patient_id']}.pt"),
                          map_location='cpu', weights_only=False)
        except: continue
        if 'mean_pred' not in d: continue
        mp = d['mean_pred'].numpy(); lab = d['label'].squeeze().numpy()
        for name,pred,true in [
            ('ET', mp[0]>cfg.SEG_THRESHOLD, lab==3),
            ('TC', mp[1]>cfg.SEG_THRESHOLD, (lab==1)|(lab==3)),
            ('WT', mp[2]>cfg.SEG_THRESHOLD, lab>=1)]:
            pred_clean = keep_largest_component(pred)   # ← added line
            res[name]['dice'].append(compute_dice(pred_clean,true))
            res[name]['hd95'].append(compute_hd95(pred_clean,true))
    seg = {}
    for r in ['WT','TC','ET']:
        dv = [x for x in res[r]['dice'] if not np.isnan(x)]
        hv = [x for x in res[r]['hd95'] if not np.isnan(x)]
        seg[r] = {'dice':np.mean(dv) if dv else float('nan'),
                  'hd95':np.mean(hv) if hv else float('nan'),
                  'dice_std':np.std(dv) if dv else 0,
                  'hd95_std':np.std(hv) if hv else 0,
                  'n':len(dv)}
    return seg


def display_all(metrics, seg, history):
    # Training curves
    fig,ax = plt.subplots(1,2,figsize=(14,5))
    ep = range(1,len(history['train_loss'])+1)
    ax[0].plot(ep,history['train_loss'],'b-o',ms=3,label='Train')
    ax[0].plot(ep,history['val_loss'],'r-o',ms=3,label='Val')
    ax[0].set_title('Loss',fontweight='bold'); ax[0].legend(); ax[0].grid(alpha=0.3)
    ax[1].plot(ep,history['train_acc'],'b-o',ms=3,label='Train')
    ax[1].plot(ep,history['val_acc'],'r-o',ms=3,label='Val')
    ax[1].set_title('Accuracy (IDH)',fontweight='bold'); ax[1].legend()
    ax[1].grid(alpha=0.3); ax[1].set_ylim([0,1])
    plt.suptitle('Training Curves',fontweight='bold'); plt.tight_layout(); plt.show()

    # Confusion matrices
    # Confusion matrices
    tasks = [t for t in ['idh','grade','mgmt'] if t in metrics]
    nm = {'idh':  'IDH\n(0=WT, 1=Mut)',
          'grade': 'Grade\n(0=Low, 1=High)',
          'mgmt':  'MGMT\n(0=Neg, 1=Pos)'}

    fig, ax = plt.subplots(1, len(tasks), figsize=(6*len(tasks), 5))
    if len(tasks) == 1:
        ax = [ax]

    for a, t in zip(ax, tasks):
        m  = metrics[t]

        # Force 2×2 regardless of collapse
        cm = confusion_matrix(m['labels'], m['preds'], labels=[0, 1])
        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=['Neg', 'Pos'])
        disp.plot(ax=a, cmap='Blues', colorbar=False)
        a.set_title(nm[t], fontweight='bold')

        auc_str = f"{m['auc']:.3f}" if not np.isnan(m['auc']) else 'N/A'
        a.set_xlabel(
            f"Acc={m['accuracy']:.3f}  F1={m['f1']:.3f}  AUC={auc_str}",
            fontsize=10)

        # Annotate collapse if it happened
        unique_preds = np.unique(m['preds'])
        if len(unique_preds) == 1:
            a.set_title(
                nm[t] + f"\n⚠ COLLAPSE (all pred={unique_preds[0]})",
                fontweight='bold', color='red')

    plt.suptitle('Confusion Matrices — UCSF-PDGM Test Set',
                fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

    # ROC
    fig,a=plt.subplots(figsize=(7,6))
    col={'idh':'#2196F3','grade':'#4CAF50','mgmt':'#FF9800'}
    for t,c in col.items():
        if t not in metrics or np.isnan(metrics[t]['auc']): continue
        m=metrics[t]; fpr,tpr,_=roc_curve(m['labels'],m['probs'])
        a.plot(fpr,tpr,color=c,lw=2.5,label=f"{m['name']} ({m['auc']:.3f})")
    a.plot([0,1],[0,1],'k--',lw=1,label='Random')
    a.set_title('ROC Curves',fontweight='bold'); a.legend(); a.grid(alpha=0.3)
    a.set_xlabel('FPR'); a.set_ylabel('TPR'); plt.tight_layout(); plt.show()

    # Segmentation bars
    regions=['WT','TC','ET']; cols=['#2196F3','#4CAF50','#FF9800']
    dv=[seg[r]['dice'] for r in regions]; hv=[seg[r]['hd95'] for r in regions]
    fig,ax=plt.subplots(1,2,figsize=(12,5))
    ax[0].bar(regions,dv,color=cols,yerr=[seg[r]['dice_std'] for r in regions],
              capsize=5,edgecolor='black',lw=0.5)
    ax[0].set_ylim([0,1.05]); ax[0].set_title('Dice (higher better)',fontweight='bold')
    ax[0].axhline(0.8,color='r',ls='--',alpha=0.5); ax[0].grid(alpha=0.3,axis='y')
    for i,v in enumerate(dv):
        if not np.isnan(v): ax[0].text(i,v+0.03,f'{v:.3f}',ha='center',fontweight='bold')
    hv2=[h if not np.isnan(h) else 0 for h in hv]
    ax[1].bar(regions,hv2,color=cols,capsize=5,edgecolor='black',lw=0.5)
    ax[1].set_title('HD95 (lower better)',fontweight='bold'); ax[1].grid(alpha=0.3,axis='y')
    for i,v in enumerate(hv):
        if not np.isnan(v): ax[1].text(i,hv2[i]+0.5,f'{v:.1f}',ha='center',fontweight='bold')
    plt.suptitle('SwinUNETR Segmentation',fontweight='bold'); plt.tight_layout(); plt.show()

    # Text summary
    print("\n"+"="*55+"\n  SUMMARY\n"+"="*55)
    print("\nClassification:")

    for t in ['idh', 'grade', 'mgmt']:    # ← 4 spaces indent
        if t not in metrics:
            continue
        m = metrics[t]
        auc = f"{m['auc']:.3f}" if not np.isnan(m['auc']) else 'N/A'
        print(f"\n{m['name']}")
        print(f"  Accuracy     : {m['accuracy']:.3f}")
        print(f"  Precision    : {m['precision']:.3f}")
        print(f"  Recall       : {m['recall']:.3f}")
        print(f"  F1           : {m['f1']:.3f}")
        print(f"  AUC          : {auc}")
        print(f"  Sensitivity  : {m['sensitivity']:.3f}")
        print(f"  Specificity  : {m['specificity']:.3f}")
        print(f"  TP={m['tp']}  FP={m['fp']}  TN={m['tn']}  FN={m['fn']}")

    print("\nSegmentation:")
    for r in ['WT','TC','ET']:
        m=seg[r]; d=f"{m['dice']:.3f}" if not np.isnan(m['dice']) else 'N/A'
        h=f"{m['hd95']:.1f}" if not np.isnan(m['hd95']) else 'N/A'
        print(f"  {r:<4} Dice={d} HD95={h}")
    print("="*55)


# ── Segmentation (deterministic, unaffected by classifier seed) ──
seg = evaluate_segmentation(CACHE_DIR, cfg.UCSF_CSV, cfg.UCSF_DIR)

# ── Aggregate classification summary across seeds (the reported numbers) ──
print("="*55)
print(f"  AGGREGATE CLASSIFICATION SUMMARY (mean ± std, n={len(SEEDS)} seeds)")
print("="*55)
for task in ['idh', 'grade', 'mgmt']:
    col = f'{task}_auc'
    if col in results_df.columns:
        print(f"\n{task.upper()}")
        for metric in ['auc', 'acc', 'f1', 'sens', 'spec']:
            c = f'{task}_{metric}'
            m, s = results_df[c].mean(), results_df[c].std()
            print(f"  {metric:<12}: {m:.3f} ± {s:.3f}")

print("\nSegmentation:")
for r in ['WT','TC','ET']:
    m = seg[r]
    print(f"  {r:<4} Dice={m['dice']:.3f}  HD95={m['hd95']:.1f}")
print("="*55)

# ── Visuals: pick one representative seed (closest to mean IDH AUC) for plots ──
rep_seed = results_df.iloc[
    (results_df['idh_auc'] - results_df['idh_auc'].mean()).abs().argsort()[:1]
]['seed'].values[0]
print(f"\nPlots below are from seed {rep_seed} (closest to mean IDH AUC) — "
      f"see aggregate numbers above for the actual reported results.")

ckpt = torch.load(os.path.join(cfg.CKPT_DIR, f'clf_best_seed{rep_seed}.pth'),
                   map_location=cfg.DEVICE, weights_only=False)
classifier.load_state_dict(ckpt['model_state'])
latest = torch.load(os.path.join(cfg.CKPT_DIR, f'clf_latest_seed{rep_seed}.pth'),
                     map_location=cfg.DEVICE, weights_only=False)
_,_,test_loader = build_loaders(SLIM_CACHE, cfg.UCSF_CSV, cfg.UCSF_DIR)
metrics_rep = evaluate_classifier(classifier, test_loader)
display_all(metrics_rep, seg, latest['history'])
```

```python
# Five-fold cross validation
from sklearn.model_selection import StratifiedKFold

def run_five_fold_cv(cache_dir, csv_path, ucsf_dir, seed=42):
    all_dicts = build_ucsf_dicts(csv_path, ucsf_dir)
    cached_pids = {f.replace('.pt','') for f in os.listdir(cache_dir)
                   if f.endswith('.pt')}
    all_dicts = [d for d in all_dicts if d['patient_id'] in cached_pids]

    # Stratify on IDH label for balanced folds
    labels = np.array([d['idh'] if not np.isnan(d['idh']) else -1
                       for d in all_dicts])
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)

    fold_results = []
    all_probs = {t: [] for t in ['idh','grade','mgmt']}
    all_labels = {t: [] for t in ['idh','grade','mgmt']}
    histories = []

    for fold, (train_idx, test_idx) in enumerate(skf.split(all_dicts, labels)):
        print(f"\n{'='*55}\n  FOLD {fold+1}/5\n{'='*55}")
        set_seed(seed + fold)

        train_dicts = [all_dicts[i] for i in train_idx]
        test_dicts  = [all_dicts[i] for i in test_idx]

        def paths(dicts):
            return [os.path.join(cache_dir, f"{d['patient_id']}.pt")
                    for d in dicts]

        train_loader = torch.utils.data.DataLoader(
            CachedDataset(paths(train_dicts)),
            batch_size=cfg.BATCH_SIZE, shuffle=True, num_workers=0)
        test_loader = torch.utils.data.DataLoader(
            CachedDataset(paths(test_dicts)),
            batch_size=1, shuffle=False, num_workers=0)

        clf = BrainTumorClassifier().to(cfg.DEVICE)
        clf, history = train_classifier(
            clf, cache_dir, csv_path, ucsf_dir, tag=f"fold{fold+1}")
        histories.append(history)

        metrics = evaluate_classifier(clf, test_loader)
        fold_results.append(metrics)

        # Collect raw probs+labels for pooled ROC
        for t in ['idh','grade','mgmt']:
            if t in metrics:
                all_probs[t].extend(metrics[t]['probs'].tolist())
                all_labels[t].extend(metrics[t]['labels'].tolist())

    # Summary across folds
    print(f"\n{'='*55}\n  5-FOLD CV SUMMARY\n{'='*55}")
    for task in ['idh','grade','mgmt']:
        aucs = [r[task]['auc'] for r in fold_results
                if task in r and not np.isnan(r[task]['auc'])]
        f1s  = [r[task]['f1']  for r in fold_results if task in r]
        print(f"\n{task.upper()}")
        print(f"  AUC  : {np.mean(aucs):.3f} ± {np.std(aucs):.3f}")
        print(f"  F1   : {np.mean(f1s):.3f} ± {np.std(f1s):.3f}")

    return fold_results, all_probs, all_labels, histories
```

```python
from statsmodels.stats.contingency_tables import mcnemar

def mcnemar_test(model_preds, baseline_preds, true_labels):
    # baseline = always predict majority class
    b = sum(1 for m, b, t in zip(model_preds, baseline_preds, true_labels)
            if m == t and b != t)  # model right, baseline wrong
    c = sum(1 for m, b, t in zip(model_preds, baseline_preds, true_labels)
            if m != t and b == t)  # model wrong, baseline right
    table = [[0, b], [c, 0]]
    result = mcnemar(table, exact=True)
    return result.pvalue
```

```python
def evaluate_segmentation_utsw(cache_dir):
    files = [f for f in os.listdir(cache_dir) if f.endswith('.pt')]
    res   = {r:{'dice':[],'hd95':[]} for r in ['ET','TC','WT']}

    for fname in tqdm(files, desc="UTSW seg eval"):
        try:
            d = torch.load(os.path.join(cache_dir, fname),
                           map_location='cpu', weights_only=False)
        except: continue
        if 'mean_pred' not in d: continue
        mp  = d['mean_pred'].numpy()
        lab = d['label'].squeeze().numpy()
        for name, pred, true in [
            ('ET', mp[0]>cfg.SEG_THRESHOLD, lab==3),
            ('TC', mp[1]>cfg.SEG_THRESHOLD, (lab==1)|(lab==3)),
            ('WT', mp[2]>cfg.SEG_THRESHOLD, lab>=1)]:
            pred_clean = keep_largest_component(pred)
            res[name]['dice'].append(compute_dice(pred_clean, true))
            res[name]['hd95'].append(compute_hd95(pred_clean, true))

    print("\n=== UTSW Cross-Dataset Segmentation (zero-shot) ===")
    for r in ['WT','TC','ET']:
        dv = [x for x in res[r]['dice'] if not np.isnan(x)]
        hv = [x for x in res[r]['hd95'] if not np.isnan(x)]
        print(f"  {r}: Dice={np.mean(dv):.3f}±{np.std(dv):.3f}  "
              f"HD95={np.mean(hv):.1f}±{np.std(hv):.1f}  (n={len(dv)})")
    return res

def evaluate_classifier_utsw(clf_model, cache_dir):
    files = [f for f in os.listdir(cache_dir) if f.endswith('.pt')]
    res   = {t: {'probs':[], 'labels':[]} for t in ['idh','grade','mgmt']}

    clf_model.eval()
    with torch.no_grad():
        for fname in tqdm(files, desc="UTSW clf eval"):
            try:
                d = torch.load(os.path.join(cache_dir, fname),
                               map_location='cpu', weights_only=False)
            except: continue
            enc  = d['encoder'].unsqueeze(0).to(cfg.DEVICE)
            mor  = d['morph'].unsqueeze(0).to(cfg.DEVICE)
            conf = d['conf_score'].view(1,1).to(cfg.DEVICE)
            preds = clf_model(enc, mor, conf)
            for t in ['idh','grade','mgmt']:
                res[t]['probs'].append(
                    torch.sigmoid(preds[t]).item())
                res[t]['labels'].append(d[t].item())

    print("\n=== UTSW Cross-Dataset Classification (zero-shot) ===")
    for t in ['idh','grade','mgmt']:
        p = np.array(res[t]['probs'])
        l = np.array(res[t]['labels'])
        valid = ~np.isnan(l)
        p, l  = p[valid], l[valid]
        if len(l) == 0:
            print(f"  {t.upper()}: no valid labels")
            continue
        try:
            auc = roc_auc_score(l, p)
        except:
            auc = float('nan')
        pb  = (p >= 0.5).astype(int)
        acc = accuracy_score(l, pb)
        f1  = f1_score(l, pb, zero_division=0)
        sens = recall_score(l, pb, zero_division=0)
        try:
            tn, fp, fn, tp = confusion_matrix(l, pb, labels=[0,1]).ravel()
            spec = tn/(tn+fp) if (tn+fp)>0 else 0.0
        except:
            spec = float('nan')
        print(f"  {t.upper()}: AUC={auc:.3f}  Acc={acc:.3f}  "
              f"F1={f1:.3f}  Sens={sens:.3f}  Spec={spec:.3f}  (n={len(l)})")
    return res

print("UTSW evaluation functions defined")
```
